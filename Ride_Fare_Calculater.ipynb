```python
def calculate_fare(distance, peak=False):         # ← 'peak' declared here
    if distance <= 5:
        base_fare = 100
    elif distance <= 10:
        base_fare = 100 + (distance - 5) * 30
    else:
        base_fare = 100 + (5 * 30) + (distance - 10) * 25

    if peak:                                      # ✅ now 'peak' exists
        surcharge = base_fare * 0.20
    else:
        surcharge = 0

    total_fare = base_fare + surcharge

    print(f"Distance: {distance} km")
    print(f"Base Fare: {base_fare:.2f}")
    print(f"Peak Hours: {'Yes' if peak else 'No'}")
    print(f"Surcharge: {surcharge:.2f}")
    print(f"Total Fare: {total_fare:.2f}")


# Top level
distance = float(input("Enter the distance: "))
calculate_fare(distance)
```
